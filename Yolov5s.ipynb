# 평가 지표 출력

```python
import sys, os
import torch
import pandas as pd
import matplotlib.pyplot as plt

# 0) YOLOv5 repo 경로(허브 캐시) 추가
repo = os.path.join(torch.hub.get_dir(), 'ultralytics_yolov5_master')
if repo not in sys.path:
    sys.path.insert(0, repo)

from val import run as val_run

# 1) 모델 로드(클래스 이름용)
weights_path = '/content/drive/MyDrive/2025 IT 경진대회_스마트 AI 포스기/0804_yolov5_training6/weights/best.pt'
data_yaml   = '/content/20class_dataset-7/data.yaml'

model = torch.hub.load('ultralytics/yolov5', 'custom', path=weights_path)
class_names = model.names

# 2) 검증 실행 → maps(클래스별 mAP@0.5) 받기
results, maps, _ = val_run(
    weights=[weights_path],
    data=data_yaml,
    imgsz=640,
    conf_thres=0.001,
    iou_thres=0.6,
    plots=False,
    verbose=False
)

# 3) DataFrame 구성 + 정렬
df_map = pd.DataFrame({
    "Class": [class_names[i] for i in range(len(maps))],
    "mAP@0.5": maps
})
df_map["mAP@0.5 (%)"] = (df_map["mAP@0.5"] * 100).round(2)
df_map = df_map.sort_values(by="mAP@0.5", ascending=False).reset_index(drop=True)

# 4) 평균 mAP 출력
mean_map = df_map["mAP@0.5"].mean() * 100
print(f"평균 mAP@0.5: {mean_map/100:.4f}  ({mean_map:.2f}%)")

# 5) 표 시각화 + 저장
fig_h = len(df_map) * 0.45 + 1.2
fig, ax = plt.subplots(figsize=(7.5, fig_h))
ax.axis("off")

table_data = df_map[["Class", "mAP@0.5 (%)"]].values.tolist()
table_data.append(["Average", round(mean_map, 2)])  # 평균 행 추가

table = ax.table(
    cellText=table_data,
    colLabels=["Class", "mAP@0.5 (%)"],
    loc="center",
    cellLoc="center"
)

# 스타일
table.auto_set_font_size(False)
table.set_fontsize(10)
table.scale(1, 1.25)

# 헤더(0행) 색
for j in range(2):
    table[(0, j)].set_facecolor("#f4a261")
    table[(0, j)].set_text_props(weight="bold", color="white")

# 평균 행 강조 (헤더가 0행이므로 평균은 len(table_data)행)
avg_row = len(table_data)
table[(avg_row, 0)].set_facecolor("#2a9d8f")
table[(avg_row, 1)].set_facecolor("#2a9d8f")
table[(avg_row, 0)].set_text_props(weight="bold", color="white")
table[(avg_row, 1)].set_text_props(weight="bold", color="white")

plt.title("Per-Class mAP@0.5 (YOLOv5)", fontsize=14, weight="bold")
plt.tight_layout()

out_path = "/content/per_class_map_table.png"
plt.savefig(out_path, dpi=200, bbox_inches="tight")
plt.show()

print(f"표 이미지 저장됨: {out_path}")
```
