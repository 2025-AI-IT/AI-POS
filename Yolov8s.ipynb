# 평가지표 mAP 출력

```python
from ultralytics import YOLO
import pandas as pd
import matplotlib.pyplot as plt

# 1) 모델 불러오기
model = YOLO("/content/drive/MyDrive/2025 IT 경진대회_스마트 AI 포스기/0730_yolov8_training2/weights/best.pt")

# 2) 검증 실행
results = model.val(data="/content/20class_dataset-6/data.yaml", split="val")  # 데이터셋 yaml 경로 수정

# 3) 클래스별 mAP@0.5
class_names = model.names
per_class_map = results.box.maps  # 리스트 형태
df_map = pd.DataFrame({
    "Class": [class_names[i] for i in range(len(per_class_map))],
    "mAP@0.5": per_class_map
})
df_map["mAP@0.5 (%)"] = (df_map["mAP@0.5"] * 100).round(2)

# 4) 평균 mAP
mean_map = df_map["mAP@0.5"].mean() * 100

# 5) 표 시각화
fig, ax = plt.subplots(figsize=(6, len(df_map) * 0.4 + 1))  # 크기 조정
ax.axis("off")  # 축 숨기기
table_data = df_map[["Class", "mAP@0.5 (%)"]].values.tolist()
table_data.append(["Average", round(mean_map, 2)])  # 마지막 줄에 평균 추가

# 테이블 생성
table = ax.table(
    cellText=table_data,
    colLabels=["Class", "mAP@0.5 (%)"],
    loc="center",
    cellLoc="center"
)

# 스타일 조정
table.auto_set_font_size(False)
table.set_fontsize(10)
table.scale(1, 1.3)

# 헤더 색상
for j in range(len(df_map.columns)-1):
    table[(0, j)].set_facecolor("#f4a261")
    table[(0, j)].set_text_props(weight="bold", color="white")

# 평균 행 색상
table[(len(table_data)-1, 0)].set_facecolor("#2a9d8f")
table[(len(table_data)-1, 1)].set_facecolor("#2a9d8f")
table[(len(table_data)-1, 0)].set_text_props(weight="bold", color="white")
table[(len(table_data)-1, 1)].set_text_props(weight="bold", color="white")

plt.title("Per-Class mAP@0.5 (YOLOv8_클래스20개)", fontsize=14, weight="bold")
plt.show()
```
